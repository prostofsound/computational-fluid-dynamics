## Radial pressure gradient in circular flow

For fluid moving along a circular streamline of radius $r$ with speed $V$, the fluid has a centripetal acceleration directed inward:

$$
a_r=-\frac{V^2}{r}
$$

Take the radial direction $r$ as positive outward. The radial pressure force per unit mass is

$$
-\frac{1}{\rho}\frac{\partial p}{\partial r}
$$

Applying Newton's second law in the radial direction:

$$
-\frac{1}{\rho}\frac{\partial p}{\partial r}
=
-\frac{V^2}{r}
$$

Therefore,

$$
\boxed{
\frac{\partial p}{\partial r}
=
\rho\frac{V^2}{r}
}
$$

## Key physical concepts

- Circular motion requires an **inward centripetal acceleration**:

$$
a_c=\frac{V^2}{r}
$$

- That inward acceleration must be produced by a net inward force.
- In an inviscid fluid, the required force comes from the **pressure gradient**.
- Since

$$
\frac{\partial p}{\partial r}>0
$$

pressure increases as radius increases:

$$
\boxed{
p_{\text{outside}}>p_{\text{inside}}
}
$$

- The higher pressure on the outside pushes the fluid inward and provides the centripetal force.
- This equation describes pressure variation **normal to the streamline**, unlike Bernoulli's equation, which describes changes primarily **along a streamline**.
- The outward increase in pressure creates the inward pressure force needed for centripetal acceleration.

For steady, incompressible, inviscid flow,

$$
\frac{p_1}{\rho}+\frac{V_1^2}{2}+gz_1
=

The pipe is horizontal, so

Rearranging,

### Step 2 — Apply continuity

For incompressible flow,

$$

$$

so

$$

and

$$

$$
\frac{2(p_1-p_2)}{\rho}
=
\left(\frac{Q}{A_2}\right)^2
-
\left(\frac{Q}{A_1}\right)^2

Factor out $Q^2$:

$$
\frac{2(p_1-p_2)}{\rho}

\left(
\frac{1}{A_2^2}-\frac{1}{A_1^2}

Factor out $1/A_2^2$:

$$
\frac{1}{A_2^2}-\frac{1}{A_1^2}

\frac{1}{A_2^2}
\left[
1-\left(\frac{A_2}{A_1}\right)^2
\right]

$$

u=6y,\qquad v=3
$$

the streamline equation is

$$
\frac{dy}{dx}=\frac{v}{u}.
$$

Therefore,

$$
\frac{dy}{dx}=\frac{3}{6y}=\frac{1}{2y}.
$$

Rearrange:

$$
2y\,dy=dx
$$

and integrate:

$$
\int 2y\,dy=\int dx
$$

so

$$
y^2=x+C.
$$

Thus the family of streamlines is

$$
\boxed{y^2=x+C}
$$

or, since the problem asks for the **upper half-plane**,

$$
\boxed{y=\sqrt{x+C}}.
$$

One physical check: because $$v=3>0$$ everywhere, the fluid always has an upward component, while $$u=6y>0$$ in the upper half-plane, so the flow moves **upward and to the right**, consistent with these streamline shapes.

```python
# Representative streamline constants
C_values = [-4, -2, 0, 2, 4]

x = np.linspace(-5, 8, 600)

plt.figure(figsize=(8, 5))

for C in C_values:
    valid = x + C >= 0
    y = np.sqrt(x[valid] + C)
    plt.plot(x[valid], y, label=fr"$C={C}$")

plt.xlabel("x [ft]")
plt.ylabel("y [ft]")
plt.title(r"Representative streamlines: $y^2=x+C$")
plt.grid(True, alpha=0.3)
plt.legend()
plt.xlim(-5, 8)
plt.ylim(0, 4)

plt.show()
```

A **streakline** is the curve formed by **all fluid particles that have passed through the same fixed point**.

Think of continuously injecting smoke or dye from one location:

> **Streakline = “Where are all the particles that came through this source point?”**

Example: smoke coming continuously from a chimney traces a streakline.

For comparison:
- **Pathline:** trajectory of one specific particle.
- **Streamline:** curve everywhere tangent to the instantaneous velocity field.
- **Streakline:** it's the shape formed by the collection of particles that passed through the same point or continuously released from the same source.

In steady flow, all three coincide.


This problem is really describing a **streakline/plume**: steam is continuously released from the fixed power plant between $t=0$ and $t=5$ h, while the wind direction changes with time.

For a particle released at time $\tau$ from

$$
\mathbf{x}_0=(-1,3)\ \text{miles},
$$

its position at time $t$ is

$$
\boxed{
\mathbf{x}(t,\tau)
=
\mathbf{x}_0+\int_{\tau}^{t}\mathbf{V}(t')\,dt'
}
$$

Since the velocity is piecewise constant, we just add displacement as

$$
\Delta\mathbf{x}=\mathbf{V}\Delta t.
$$

### At $t=3$ h

The wind is

$$
\mathbf V_1=(10,-5)\ \text{mph}.
$$

The newest particle is at the plant:$(-1,3)$

while the particle released at $t=0$ has traveled for 3 h:

$$
(-1,3)+(10,-5)(3)
=
(29,-12).
$$

So the plume is the straight segment

$$
\boxed{(-1,3)\rightarrow(29,-12)}.
$$

### At $t=10$ h

The steam release stopped at $t=5$ h.

The particle released at $t=5$ experiences the second wind for 5 h:

$$
(-1,3)+(15,8)(5)
=
\boxed{(74,43)}.
$$

The particle released at $t=3$ experiences it for 7 h:

$$
(-1,3)+(15,8)(7)
=
\boxed{(104,59)}.
$$

The oldest particle, released at $t=0$, first travels 3 h with the first wind and 7 h with the second:

$$
(-1,3)+(10,-5)(3)+(15,8)(7)
$$

$$
=
\boxed{(134,44)}.
$$

Thus the plume is

$$
\boxed{
(74,43)\rightarrow(104,59)\rightarrow(134,44)
}
$$

which explains the bend in the plotted line.

### At $t=15$ h

From $t=10$ to $t=15$,

$$
\mathbf V_3=(5,0)\ \text{mph}.
$$

Every particle moves an additional

$$
(5,0)(5)=(25,0)\ \text{miles}.
$$

Therefore the whole $t=10$ plume shifts 25 miles to the right:

$$
\boxed{
(99,43)\rightarrow(129,59)\rightarrow(159,44)
}
$$

So the key result is:

$$
\boxed{
t=3:\;(-1,3)\rightarrow(29,-12)
}
$$

$$
\boxed{
t=10:\;(74,43)\rightarrow(104,59)\rightarrow(134,44)
}
$$

$$
\boxed{
t=15:\;(99,43)\rightarrow(129,59)\rightarrow(159,44)
}
$$

The graph above is drawn to scale. Notice that at $t=15$ the shape no longer changes because all particles are experiencing the same horizontal velocity; the entire plume simply translates to the right.

```python
source = np.array([-1.0, 3.0])  # miles

V1 = np.array([10.0, -5.0])    # mph, 0 < t < 3 h
V2 = np.array([15.0,  8.0])    # mph, 3 < t < 10 h
V3 = np.array([ 5.0,  0.0])    # mph, t > 10 h

# Plume at t = 3 h: releases from tau = 0 to 3 h
p3_oldest = source + V1 * 3
p3_newest = source

# Plume at t = 10 h
p10_tau5 = source + V2 * (10 - 5)                   # particle released at 5 h
p10_tau3 = source + V2 * (10 - 3)                   # particle released at 3 h
p10_tau0 = source + V1 * 3 + V2 * (10 - 3)          # particle released at 0 h

# Plume at t = 15 h: all t=10 positions are shifted by V3*(15-10)
shift_10_15 = V3 * 5
p15_tau5 = p10_tau5 + shift_10_15
p15_tau3 = p10_tau3 + shift_10_15
p15_tau0 = p10_tau0 + shift_10_15

plt.figure(figsize=(9, 6))

plt.plot([p3_newest[0], p3_oldest[0]],
         [p3_newest[1], p3_oldest[1]],
         marker='o', label='t = 3 h')

plt.plot([p10_tau5[0], p10_tau3[0], p10_tau0[0]],
         [p10_tau5[1], p10_tau3[1], p10_tau0[1]],
         marker='o', label='t = 10 h')

plt.plot([p15_tau5[0], p15_tau3[0], p15_tau0[0]],
         [p15_tau5[1], p15_tau3[1], p15_tau0[1]],
         marker='o', label='t = 15 h')

plt.scatter(source[0], source[1], marker='x', s=90, label='Power plant')

plt.xlabel('x [miles]')
plt.ylabel('y [miles]')
plt.title('Radioactive steam plume at selected times')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("t = 3 h endpoints:", p3_newest, p3_oldest)
print("t = 10 h key points:", p10_tau5, p10_tau3, p10_tau0)
print("t = 15 h key points:", p15_tau5, p15_tau3, p15_tau0)
```

```python
# Time interval
t = np.linspace(0, 3, 1200)

# Dust devil motion
x = 5*t + 10*np.cos(2*np.pi*t)
y = 10*np.sin(2*np.pi*t)

plt.figure(figsize=(9, 5))
plt.plot(x, y)
plt.scatter([x[0]], [y[0]], s=50, label="Start, t = 0 s")
plt.scatter([x[-1]], [y[-1]], s=50, label="End, t = 3 s")

plt.xlabel("x [ft]")
plt.ylabel("y [ft]")
plt.title("Pathline of the fluid particle")
plt.axis("equal")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

This is a **pathline** problem because we are following one specific fluid particle as the dust devil moves.

The particle has two motions at the same time:

1. Circular motion around the dust-devil center:

$$
x_r=10\cos(2\pi t)
$$

$$
y_r=10\sin(2\pi t)
$$

The term $2\pi t$ comes from one revolution per second:

$$
\omega=2\pi\ \text{rad/s}.
$$

2. Translation of the whole dust devil in the positive $x$-direction at

$$
U=5\ \text{ft/s},
$$

so

$$
x_T=5t.
$$

Add the two motions:

$$
\boxed{x(t)=5t+10\cos(2\pi t)}
$$

$$
\boxed{y(t)=10\sin(2\pi t)}
$$

for

$$
\boxed{0\leq t\leq3\ \text{s}}.
$$

So the particle makes three revolutions while the whole vortex moves $5(3)=15\ \text{ft}$ to the right.

The resulting pathline is the looped curve shown above.

The important physical point is that the rotational speed is much larger than the translation speed:

$$
V_{\text{rot}}=\omega r=(2\pi)(10)\approx62.8\ \text{ft/s}
$$

versus only

$$
V_{\text{translation}}=5\ \text{ft/s},
$$

which is why the trajectory forms large overlapping loops rather than just a wavy line.

This problem is testing the difference between **local acceleration** and **convective acceleration**.

For a 1D flow, the total acceleration is

$$
\boxed{
a=\frac{\partial V}{\partial t}
+
V\frac{\partial V}{\partial x}
}
$$

The first term is the **local acceleration**; the second is the **convective acceleration**.

### 1. Local acceleration at point (1)

Given

$$
V_1=0.50t
$$

differentiate with respect to time while staying at point (1):

$$
a_{\text{local},1}
=
\frac{\partial V_1}{\partial t}
=
0.50
$$

Therefore,

$$
\boxed{
a_{\text{local},1}=0.50\ \text{m/s}^2
}
$$

### 2. Local acceleration at point (2)

Similarly,

$$
V_2=1.0t
$$

so

$$
a_{\text{local},2}
=
\frac{\partial V_2}{\partial t}
=
1.0
$$

Therefore,

$$
\boxed{
a_{\text{local},2}=1.0\ \text{m/s}^2
}
$$

### 3. Sign of the convective acceleration

At any time $t>0$,

$$
V_2>V_1.
$$

So the fluid velocity **increases as we move downstream** from (1) to (2):

$$
\frac{\partial V}{\partial x}>0.
$$

Since the flow is also in the positive $x$ direction,

$$
V>0.
$$

Thus,

$$
a_{\text{conv}}
=
V\frac{\partial V}{\partial x}>0.
$$

Therefore,

$$
\boxed{\text{The average convective acceleration is positive.}}
$$

Physically, there are **two reasons a fluid particle accelerates here**: the flow at every fixed location is getting faster with time (**local acceleration**), and the pipe contracts, so a particle encounters higher velocities as it moves downstream (**convective acceleration**).
